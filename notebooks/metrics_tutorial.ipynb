### Compute metrics on detections from a object-detection network 📈

This notebook is a simple demo on how to use the `metrics.py` module in the "utils" folder.

```python
%cd ..
```

If you have not already, you will need a generated neural network using the `./generate` command.

NOTE: Particularly, we need the modules `input_preparator.py` and `output_preparator.py` that are generated with the network itself and inside the generated networks' folder. Without those modules, we will not be able to perform the metrics computation.

```python
%%bash
./generate networks/object-detection/yolov5s6-relu/onnx/network_f16.yaml -d yolov5s6 -f
```

Perform the metrics computation with the next command. The arguments and syntax are:

`python utils/metrics.py <gen_dir> --dataset=<dataset> --device=<device>`

- `<gen_dir>`: the relative path to the generated network.
- `<dataset>`: the name of the dataset to use (coco8, coco128 or coco).
- `<device>`: by default mppa, but can also be cpu for reference.

We will start gently with the smallest dataset: coco8.

```python
%%bash
# We need to source inside the same cell we execute the metrics script
# source <kalray toolchain>
python utils/metrics.py yolov5s6 --dataset=coco8 --device=mppa
```

The interpretation of this result is as follows:
- P: precision
- R: recall
- mAP50: mean average precision on a confidence threshold of 0.5.
- mAP50-95: mean average precision on a confidence threshold from 0.5 to 0.95 in steps of 0.05.

Read more about these metrics here: https://docs.ultralytics.com/fr/guides/yolo-performance-metrics/

If the last metrics computation worked, we can move to the second biggest dataset: coco128.

```python
%%bash
# We need to source inside the same cell we execute the metrics script
# source <kalray toolchain>
python utils/metrics.py yolov5s6 --dataset=coco128 --device=mppa
```

And finally, we can test the biggest dataset, named just coco, containing 5000 images. This will take around 6 minutes on mppa.

NOTE: this will stress your mppa for the inferencce step, then you RAM for loading the results from memory, and finally your CPU to compute the metrics and store their results. If too much RAM is used, your host device might malfunction or directly crash. There is a chunking mecanism that should avoid this to happen, but we do not recommend using this program if you are already consuming 40% of your RAM.

```python
%%bash
# We need to source inside the same cell we execute the metrics script
# source <kalray toolchain>
python utils/metrics.py yolov5s6 --dataset=coco --device=mppa
```
